```python
#Kegiatan 1 — NumPy: Operasi Vektor (±50 menit)

import numpy as np                     # np = nama panggilan standar

uts = np.array([78, 90, 66, 85, 72])   # list -> array NumPy
uas = np.array([80, 94, 52, 88, 75])

nilai_akhir = 0.4 * uts + 0.6 * uas    # rumus dihitung SEKALIGUS utk 5 mhs
print("Nilai akhir :", nilai_akhir)
print("Rata-rata   :", nilai_akhir.mean())
print("Tertinggi   :", nilai_akhir.max())
```

```python
#PENGEMBANGAN - Kegiatan 1 (Menghitung rata-rata, data hasil panen tertinggi, dan data hasil panen terendah menggunakan NumPy)

import pandas as pd
import numpy as np

# Membaca dataset
df = pd.read_csv("yield_df.csv")

# Mengambil data hasil panen
hasil_panen = np.array(df["hg/ha_yield"])

# Menghitung rata-rata, tertinggi, dan terendah
rata_rata = np.mean(hasil_panen)
hasil_tertinggi = np.max(hasil_panen)
hasil_terendah = np.min(hasil_panen)

print("Rata-rata hasil panen :", round(rata_rata, 2), "hg/ha")
print("Hasil panen tertinggi :", hasil_tertinggi, "hg/ha")
print("Hasil panen terendah  :", hasil_terendah, "hg/ha")
```

**PENJELASAN :**  
Praktikum kegiatan 1 ini menggunakan NumPy untuk melakukan operasi vektor. Contohnya adalah menghitung nilai akhir mahasiswa dari nilai UTS dan UAS dengan bobot tertentu. NumPy digunakan karena operasi dapat dilakukan secara sekaligus pada seluruh data tanpa harus menggunakan perulangan for. Dan juga memperkenalkan fungsi statistik seperti mean() untuk rata-rata dan max() untuk mencari nilai tertinggi.

Bagian pengembangan:  
Pada pengembangan, konsep operasi vektor tersebut diterapkan di pengembangan kegiatan 1 pada dataset hasil panen pertanian. Data hasil panen pada kolom hg/ha_yield diubah menjadi array NumPy, kemudian dihitung nilai rata-rata, hasil panen tertinggi, dan hasil panen terendah. Jadi, konsepnya tetap sama di kegiatan 1, yaitu menggunakan NumPy untuk mengolah data angka, tetapi data dan kasusnya dikembangkan dari nilai mahasiswa menjadi data hasil panen pertanian.

```python
#PENGEMBANGAN - Kegiatan 2 (Mengolah data hasil panen dan membandingkan produktivitas tanaman)

import pandas as pd

# Membaca dataset
df = pd.read_csv("yield_df.csv")

# Menampilkan 5 data pertama
print("Data awal:")
print(df.head())

# Menghitung rata-rata hasil panen
rata_rata = df["hg/ha_yield"].mean()

# Memfilter data yang hasil panennya di atas rata-rata
di_atas_rata = df[df["hg/ha_yield"] > rata_rata]

# Menghitung rata-rata hasil panen berdasarkan jenis tanaman
produktivitas = df.groupby("Item")["hg/ha_yield"].mean()
produktivitas = produktivitas.sort_values(ascending=False)

print("\nRata-rata hasil panen :", round(rata_rata, 2), "hg/ha")

print("\nContoh data dengan hasil panen di atas rata-rata:")
print(di_atas_rata[["Area", "Item", "Year", "hg/ha_yield"]].head(10))

print("\nRata-rata produktivitas setiap tanaman:")
print(produktivitas)
```

**PENJELASAN:**

Pada kegiatan 2 ini menggunakan Pandas untuk membaca dan mengolah file CSV. Data dibaca menggunakan pd.read_csv(), kemudian head() digunakan untuk melihat lima data pertama, info() untuk melihat informasi kolom dan data kosong, serta describe() untuk melihat statistik data. Dan juga mengajarkan filter data, membuat kolom baru, dan groupby() untuk menghitung rata-rata berdasarkan kelompok tertentu.

Bagian pengembangan:  
Pada pengembangan, penerapan konsep tersebut digunakan pada dataset hasil panen. Setelah data dibaca, lima data pertama ditampilkan untuk mengetahui bentuk dataset. Kemudian dihitung rata-rata hasil panen dan dilakukan filter untuk mencari data yang hasil panennya berada di atas rata-rata. Setelah itu, data dikelompokkan berdasarkan jenis tanaman menggunakan groupby() untuk mengetahui rata-rata produktivitas setiap tanaman. Dengan begitu, pengembangan tidak hanya membaca data, tetapi juga membandingkan produktivitas antarjenis tanaman.

```python
#PENGEMBANGAN - Kegiatan 3 (Membuat visualisasi hasil panen dan hubungan faktor pertanian)

import pandas as pd
import matplotlib.pyplot as plt

# Membaca dataset
df = pd.read_csv("yield_df.csv")

# Grafik 1: rata-rata hasil panen berdasarkan tahun
hasil_tahunan = df.groupby("Year")["hg/ha_yield"].mean()

plt.figure(figsize=(10, 5))
plt.plot(hasil_tahunan.index, hasil_tahunan.values, marker="o")

plt.title("Rata-rata Hasil Panen Berdasarkan Tahun")
plt.xlabel("Tahun")
plt.ylabel("Hasil Panen (hg/ha)")
plt.grid(True)
plt.show()


# Grafik 2: hubungan faktor pertanian (curah hujan) dengan hasil panen
plt.figure(figsize=(8, 5))
plt.scatter(
    df["average_rain_fall_mm_per_year"],
    df["hg/ha_yield"],
    alpha=0.5
)

plt.title("Hubungan Curah Hujan dengan Hasil Panen")
plt.xlabel("Curah Hujan (mm/tahun)")
plt.ylabel("Hasil Panen (hg/ha)")
plt.show()
```

PENJELASAN:  
Praktikum kegiatan 3 ini memperkenalkan Matplotlib untuk membuat visualisasi data. Contohnya menggunakan histogram untuk melihat sebaran nilai dan scatter plot untuk melihat hubungan antara dua data numerik. Juga dijelaskan bahwa grafik dipilih berdasarkan tujuan analisis, misalnya grafik garis untuk melihat perubahan dari waktu ke waktu dan scatter plot untuk melihat hubungan antara dua variabel.

Bagian pengembangan:  
Pada bagian pengembangan, konsep visualisasi tersebut diterapkan pada data hasil panen pertanian. Grafik pertama menggunakan grafik garis untuk melihat perubahan rata-rata hasil panen berdasarkan tahun. Kemudian dibuat scatter plot untuk melihat hubungan antara curah hujan dengan hasil panen. Curah hujan digunakan karena pada rancangan pengembangan memang terdapat bagian hubungan faktor pertanian, dan dataset yang digunakan memiliki kolom average_rain_fall_mm_per_year. Jadi, pengembangan Kegiatan 3 tidak hanya menampilkan data hasil panen, tetapi juga mengamati hubungan hasil panen dengan salah satu faktor pertanian.

**Latihan 2**

Soal:  
Tambahkan kolom status berisi LULUS/REMIDI (batas 60 pada nilai_akhir), lalu
tampilkan berapa mahasiswa REMIDI per kelas (petunjuk: groupby + filter).

```python
# Menambahkan kolom status
df["status"] = df["nilai_akhir"].apply(
    lambda x: "LULUS" if x >= 60 else "REMIDI"
)

# Memfilter mahasiswa yang REMIDI
remidi = df[df["status"] == "REMIDI"]

# Menghitung jumlah mahasiswa REMIDI per kelas
jumlah_remidi = remidi.groupby("kelas").size()

print("Jumlah mahasiswa REMIDI per kelas:")
print(jumlah_remidi)
```
